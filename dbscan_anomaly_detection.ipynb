# DBSCAN Anomaly Detection

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

### Create Dataset

```python
from sklearn.datasets import make_circles

X, y = make_circles(n_samples=1000, factor=0.3, noise=0.1, random_state=43)
```

```python
X
```

#### Visualization Initial Scatter

```python
plt.scatter(X[:,0],X[:,1], c=y)
```

### DBSCAN Model Training

```python
from sklearn.cluster import DBSCAN

dbs = DBSCAN(eps=0.10)
```

```python
predicted_data = dbs.fit_predict(X)
```

```python
predicted_data
```

```python
value = []
for i in predicted_data:
    if i not in value:
        value.append(i)
print(value)
```

```python
np.unique(predicted_data)
```

Observation: -1 are the outliers

### Visualization with DBSCAN Clustering

```python
plt.scatter(X[:,0], X[:,1], c=dbs.labels_)
```
